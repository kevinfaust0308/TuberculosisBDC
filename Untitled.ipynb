country: 219 countries

g_whoregion: EMR = eastern mediterranean, EUR = europe, AFR = africa, WPR = western pacific, AMR = america, SEA = south-east asia

year: 16 year ranges

e_pop_num: estimated population 

source_inc: method to derive incidence estimates (all forms)
source_mort: method to derive mortality estimates (all forms)

#### _100K is per 100k people
#### _LO is low bound
#### _HI is high bound

### THIS STUFF IS ONLY FOR 2015
#### _F/M, is female/male
#### _014 is between 0-14 age
#### _15PLUS is older than 15 age

#### F/M014 is female/male between 0-14 age,
#### _F/M15PLUS is female/male older than 15 age

## INCIDENCE REPORTS:

### general case counts
e_inc_num: estimated number of tuberculosis cases (all forms)
(all of above)

### tb + hiv
e_tbhiv_prct: estimated percent of tuberculosis cases involving HIV
(lo, hi)

e_inc_tbhiv_num: estimated number of tuberculosis cases that are HIV positive
(100k, lo, hi)

### mortality
e_mort_num: estimated mortality of tuberculosis cases (all forms)
(100k, lo, hi)

e_mort_exc_tbhiv_num: estimated mortality of tuberculosis cases (excluding hiv)
(100k, lo, hi)

e_mort_tbhiv_num: estimated mortality of tuberculosis cases that are HIV postive
(100k, lo, hi)

## other stuff (still relating to incedences)
cfr: estimated tuberculosis cases fatality ratio (ONLY 2015)
(lo, hi)

c_newinc: total of new and relapse cases and cases with unknown previous TB treatment history

c_cdr: tuberculosis detection rate % (all forms)

# World map 2015 with tuberculosis cases 

```python
columns = ['country', 'iso3', 'e_pop_num', 'e_inc_num']
year = 2015
x=df[df['year']==year][columns].groupby(by=['country','iso3']).mean()
x=x.reset_index(level=['country', 'iso3'])
```

```python
data = [ dict(
        type = 'choropleth',
        locations = x['iso3'],
        z = x['e_inc_num'],
        text = x['country'],
        colorscale = [[0,"rgb(5, 10, 172)"],[0.35,"rgb(40, 60, 190)"],[0.5,"rgb(70, 100, 245)"],\
            [0.6,"rgb(90, 120, 245)"],[0.7,"rgb(106, 137, 247)"],[1,"rgb(220, 220, 220)"]],
        autocolorscale = False,
        reversescale = True,
        marker = dict(
            line = dict (
                color = 'rgb(180,180,180)',
                width = 0.5
            ) ),
        colorbar = dict(
            title = 'Incidences'),
      ) ]

layout = dict(
    title = '2015 Tuberculosis Cases. Source: WHO',
    geo = dict(
        showframe = False,
        showcoastlines = False,
        projection = dict(
            type = 'Mercator'
        )
    )
)

fig = dict( data=data, layout=layout )
py.iplot(fig)
```

# Year vs average occurence around the world

```python
columns = ['year', 'e_pop_num', 'e_inc_num', 'e_mort_num']

x=df[columns].groupby(by='year').mean()
x=x.reset_index('year')
```

```python
year = x['year']
incidences = x['e_inc_num']
mort = x['e_mort_num']

trace1 = go.Bar(
    x=year,
    y=incidences,
    name='Incidences',
    marker=dict(
        color='rgb(55, 83, 109)'
    )
)
trace2 = go.Bar(
    x=year,
    y=mort,
    name='Mortality',
    marker=dict(
        color='rgb(26, 118, 255)'
    )
)
data = [trace1, trace2]
layout = go.Layout(
    title='Incidences and mortality',
    xaxis=dict(
        tickfont=dict(
            size=14,
            color='rgb(107, 107, 107)'
        )
    ),
    yaxis=dict(
        title='Average count',
        titlefont=dict(
            size=16,
            color='rgb(107, 107, 107)'
        ),
        tickfont=dict(
            size=14,
            color='rgb(107, 107, 107)'
        )
    ),
    legend=dict(
        x=0,
        y=1.0,
        bgcolor='rgba(255, 255, 255, 0)',
        bordercolor='rgba(255, 255, 255, 0)'
    ),
    barmode='group',
    bargap=0.15,
    bargroupgap=0.1
)

fig = go.Figure(data=data, layout=layout)
py.iplot(fig)
```

# data exploration

## by region

```python
# average number of total TB cases (over 16 years) by region
columns = ['g_whoregion', 'e_pop_num', 'e_inc_num', 'e_mort_num']
k=df[columns].groupby(by='g_whoregion').mean().sort_values(by='e_inc_num',ascending=False)
k

# south-east asia has most population and occurence
# more TB occurences = more deaths
```

## by country

```python
# average number of total tuberculosis cases (over 16 years) per country

columns = ['country', 'e_pop_num', 'e_inc_num', 'e_mort_num']
df[columns].groupby(by='country').mean().sort_values(by='e_inc_num',ascending=False).head()

# densely populated developing countries have the most occurences
# high mortality
```

## by years

```python
# tb across the years

columns = ['year', 'e_pop_num', 'e_inc_num', 'e_mort_num']

df[columns].groupby(by='year').mean()

# case count increased until 2006 and then is decreasing
# mortality rates decreasing
# likely due to advances in medicine
```

## 2015 specific data

```python
columns = ['g_whoregion', 'e_inc_num_m', 'e_inc_num_f']

df[df['year']==2015][columns].groupby(by='g_whoregion').mean().grouby(by=['e_inc_num_m','e_inc_num_f']).mean()
```
